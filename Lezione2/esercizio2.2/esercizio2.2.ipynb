```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import math

def f(k,x):
    return k*math.sqrt(x)
```

### Caso discreto su tre assi
Andamento di $\sqrt{\langle |\vec{r}_N|^2 \rangle_{RW}}$ in funzione del numero di passi 'steps'.

```python
x, f1, error1 = np.loadtxt("Distanza.txt", usecols=(0,1,2), delimiter=' ', unpack=True)

k = 0.999803

fit_val = np.zeros(len(x))
for i in range(len(x)):
    fit_val[i]=f(k,x[i])

plt.figure(figsize=(10,6))
plt.errorbar(x,f1,yerr=error1, label="Random walk values")
plt.plot(x, fit_val, label="Fit")
plt.xlabel('steps')
plt.ylabel(r'$\sqrt{\langle |\vec{r}_N|^2 \rangle}$')
plt.legend()
plt.title("Random walk on a cubic lattice")
plt.grid(True)

plt.show()
```

### Commenti
La simulazione riguarda $M=10^4$ random walk fatti evolvere per 100 steps.\
Il primo grafico si riferisce al caso in cui ciascuno step viene eseguito lungo una qualsiasi delle tre direzioni assiali e per ciascuna direzione in uno qualsiasi dei due versi. Il secondo grafico si riferisce invece al caso continuo, in cui vengono estratti due angoli: $\phi \in [0, 2\pi]$ in modo uniforme, e $\theta \in [0, \pi]$ calcolando la cumulativa e invertendola: 
$$p(x) = \frac{1}{2} \sin(\theta)$$
$$y = \int_0^x \frac{1}{2} \sin(\theta)d\theta = \frac{1}{2}(1-\cos(x))$$
$$x = \arccos(1-2y)$$
Con $y$ estratta uniformemente in $[0,1]$.\
Entrambi i grafici rispecchiano un andamento diffusivo risultando compatibili con la legge $f(x) = k \sqrt x$ con $k$ che è stato ricavato da un fit eseguito con gnuplot e si avvicina molto a 1 con un errore molto basso.

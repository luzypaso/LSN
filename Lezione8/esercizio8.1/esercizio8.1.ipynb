```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import math
```

### Ground state energy through Variational Monte Carlo

```python
x, func, err = np.loadtxt("Punto1.dat", usecols=(0,1,2), unpack='true')
E0 = -0.46046587969271563

plt.errorbar(x, func, yerr=err, label="Variational MC")
plt.axhline(y=E0,color="red",linestyle="--",label="Theoretical value")
plt.xlabel("Blocks")
plt.ylabel("$\\langle\\hat{H}\\rangle$")
plt.title("Ground state energy")
plt.grid(True)
plt.legend()
plt.show()
```

### Commenti
Dal grafico si osserva come il valore ottenuto con il metodo variazionale di Monte Carlo si avvicini molto al valore teorico del ground state $E_0 = -0.460$. E' stato scelto uno step size tale da avere una accettanza di circa il 50%.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

### Dado uniforme

```python
N = [1, 2, 10, 100]

all_data = np.loadtxt("Uniform.txt")

fig, axs = plt.subplots(2,2, figsize=(10,8))
axs = axs.flatten()

for i in range(4):

    data = all_data[:,i]

    axs[i].hist(data,
                bins=50,
                edgecolor='blue',
                range=(0,1),
                density=True)

    axs[i].set_title(f"N = {N[i]}")
    axs[i].set_xlabel("$S_N$")
    axs[i].set_ylabel("Frequenza")
    axs[i].grid(True, alpha=0.3)
    axs[i].set_xlim(0,1)

    if i == 3:
        dev_st = np.std(data)
        x = np.linspace(0,1,150)
        y = norm.pdf(x,0.5,dev_st)
        axs[i].plot(x,y)

plt.tight_layout()
plt.show()
```

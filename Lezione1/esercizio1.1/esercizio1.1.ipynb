```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

### Plot $\langle r \rangle$
Calcolo del valore di $\langle r \rangle = \int_0^1 r dr$ e della sua incertezza sottraendo il risultato esatto di $1/2$, in funzione degli N blocchi.

```python
x1, f1, error1 = np.loadtxt("Punto1.1.txt", usecols=(0,1,2), delimiter=' ', unpack=True)
plt.errorbar(x1,f1,yerr=error1)
plt.xlabel('N blocks')
plt.ylabel(r'$\langle r \rangle$')
plt.grid(True)

plt.show()
```

### Plot $\langle \sigma^{2} \rangle$
Calcolo del valore di $\sigma^2 = \int_0^1 (r-1/2)^2 dr$ e cioè di $\langle (r-1/2)^2 \rangle$ e della sua incertezza sottraendo il risultato esatto di $1/12$, in funzione degli N blocchi.

```python
x2, f2, error2 = np.loadtxt("Punto1.2.txt", usecols=(0,1,2), delimiter=' ', unpack=True)
plt.errorbar(x2,f2,yerr=error2)
plt.xlabel('N blocks')
plt.ylabel(r'$\sigma^2 = \langle (r-1/2)^2 \rangle$')
plt.grid(True)

plt.show()
```

### Plot $\chi^{2}$ distribution
Calcolo del $\chi^2$ di una distribuzione uniforme di valori compresi tra [0,1), ripetuto 100 volte.

```python
file = open("./Punto1.3.txt","r")
N = int(10000)
M = int(100)
chi2exp = np.zeros(N)
for i in range(N):
    chi2exp[i]=float(file.readline())

from scipy.stats import chi2
x = np.linspace(50,160)
pdf = chi2.pdf(x,M)         # funzione che restituisce la pdf del chi2: i parametri sono (intervallo di ascisse, gradi di libertà)
plt.hist(chi2exp,M,density=True)
plt.plot(x,pdf)
plt.grid(True)
plt.xlabel("$\\chi^2$")
plt.show()
```

### Commenti
Sia i valori di $\langle r \rangle$ che di $\langle \sigma^{2} \rangle$ sono compatibili con i risultati esatti.\
La distribuzione del $\chi^{2}$ è compatibile con quella teorica utilizzando 100 GdL corrispondenti a $n/M$ ($n$ random numbers e $M$ intervalli).

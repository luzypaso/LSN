```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import math
```

### Stima di $\pi$
Si è stimato il valore di $\pi$ tramite l'esperimento di Buffon. Ogni blocco contiene una stima di $\pi$ fatta su 1000 lanci e ciò che viene rappresentato è l'andamento del valor medio al crescere del numero di blocchi, da 1 a 100, insieme alla relativa incertezza.

```python
x1, f1, error1 = np.loadtxt("Buffon.txt", usecols=(0,1,2), delimiter=' ', unpack=True)
plt.errorbar(x1,f1,yerr=error1)
plt.xlabel('N blocks')
plt.ylabel(r'$\pi$')
plt.grid(True)
plt.axhline(y=math.pi, color='r')

plt.show()
```

### Commenti
Per simulare questo esperimento sono stati estratti casualmente $x$ (posizione del centro della barretta rispetto alla linea) tra $[0, d/2]$ e $x_2$ (valore assoluto della proiezione, ortogonale alle linee, di una estremità rispetto al centro) tra $[0, l/2]$. Il valore di $x$ è stato estratto in questo intervallo perché la distanza del centro dalla linea **più vicina** può assumere solo questi valori; di conseguenza in questo sistema di riferimento la linea si trova alla posizone 0 e per la simmetria del problema tutto il sistema si può ricondurre a questo intervallo di valori. Quindi, la barretta interseca la linea solo se $x_2 \geq x$, ovvero quando $x - x_2$, corrispondente all'estremo inferiore, è $\leq 0$ e quindi pari o oltre la linea più vicina.\
Per evitare di utilizzare il valore di $\pi$ all'interno del codice, lo si è stimato dal rapporto tra l'area di un quadrato e l'area del cerchio inscritto, pari a $4/\pi$. Come si può notare dal grafico, i risultati della simulazione sono comptabili con il valore teorico.

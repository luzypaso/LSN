```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

### Simulazione di 5000 steps

```python
temp = np.loadtxt("../OUTPUT/temperature5000.dat", usecols=(1), skiprows=1, unpack='true')

x = np.arange(1,51)
blocks = x.size

fig, axs = plt.subplots(1, 2, figsize=(14, 6))

axs[0].plot(x,temp[:50],markersize=4, label="Direct")
axs[0].plot(x+blocks,temp[50:100],markersize=4, label="Inverse")
axs[0].set_ylabel('Temperature')
axs[0].set_xlabel('Blocks')
axs[0].legend()
axs[0].set_title("Temperature")
axs[0].grid(True)

pot = np.loadtxt("../OUTPUT/potential_energy5000.dat", usecols=(1), skiprows=1, unpack='true')

axs[1].plot(x,pot[:50],markersize=4, label="Direct")
axs[1].plot(x+blocks,pot[50:100],markersize=4, label="Inverse")
axs[1].set_ylabel('Potential energy')
axs[1].set_xlabel('Blocks')
axs[1].legend()
axs[1].set_title("Potential Energy")
axs[1].grid(True)

fig.suptitle("Simulazione di 5000 steps", fontsize=20)

plt.show()
```

### Simulazione di 50000 steps

```python
temp = np.loadtxt("../OUTPUT/temperature50000.dat", usecols=(1), skiprows=1, unpack='true')

x = np.arange(1,51)
blocks = x.size

fig, axs = plt.subplots(1, 2, figsize=(14, 6))

axs[0].plot(x,temp[:50],markersize=4, label="Direct")
axs[0].plot(x+blocks,temp[50:100],markersize=4, label="Inverse")
axs[0].set_ylabel('Temperature')
axs[0].set_xlabel('Blocks')
axs[0].legend()
axs[0].set_title("Temperature")
axs[0].grid(True)

pot = np.loadtxt("../OUTPUT/potential_energy50000.dat", usecols=(1), skiprows=1, unpack='true')

axs[1].plot(x,pot[:50],markersize=4, label="Direct")
axs[1].plot(x+blocks,pot[50:100],markersize=4, label="Inverse")
axs[1].set_ylabel('Potential energy')
axs[1].set_xlabel('Blocks')
axs[1].legend()
axs[1].set_title("Potential Energy")
axs[1].grid(True)

fig.suptitle("Simulazione di 50000 steps", fontsize=20)

plt.show()
```

### Commenti
Per quanto riguarda la simulazione corta, condotta con $5 \cdot 10^3$ steps, l'inversione temporale mostra un andamento delle proprietà del sistema (temperatura ed en. pot.) speculare rispetto alla simulazione diretta, riportando il sistema nella condizione iniziale di bassa entropia, e mostrando la reversibilità dell'algoritmo di Verlet per un numero di passi pari ad almeno $5 \cdot 10^3$.

La seconda simulazione, invece, mostra che il sistema, una volta giunto nella configurazione ad alta entropia, non riesce più a tornare nella configurazione iniziale e continua ad evolvere in modo da mantenere la condizione di equilibrio raggiunta. Ciò è dovuto al fatto che l'algoritmo ha accumulato troppi errori numerici di arrotondamento avendo compiuto $5 \cdot 10^4$ steps, e di conseguenza non riesce più a tornare indietro.

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

### Computation at time $t=0$ **via Monte Carlo** the European call-option price $C[S(0),0]$
In blue by sampling **directly** the final asset price $S(T)$.\
In orange by sampling the **discretized $GBM(r,\sigma^2)$ path** of the asset price, dividing $[0,T]$ in $100$ time intervals.

```python
x, f1, error1 = np.loadtxt("DirectCall.txt", usecols=(0,1,2), delimiter=' ', unpack=True)
f2, error2 = np.loadtxt("DiscretizedCall.txt", usecols=(1,2), delimiter=' ', unpack=True)
plt.figure(figsize=(8,6))
plt.errorbar(x,f1,yerr=error1, label="Direct method")
plt.errorbar(x,f2,yerr=error2, label="Discretized method")
plt.xlabel('N blocks')
plt.ylabel('Call')
plt.axhline(y=14.975790778311286, linestyle="--", color="r", label="Theoretical value")
plt.title("Computation of the European call-option price")
plt.grid(True)
plt.legend()

plt.show()
```

### Computation at time $t=0$ **via Monte Carlo** the European put-option price $P[S(0),0]$:
In blue by sampling **directly** the final asset price $S(T)$.\
In orange by sampling the **discretized $GBM(r,\sigma^2)$ path** of the asset price, dividing $[0,T]$ in $100$ time intervals.

```python
x, f1, error1 = np.loadtxt("DirectPut.txt", usecols=(0,1,2), delimiter=' ', unpack=True)
f2, error2 = np.loadtxt("DiscretizedPut.txt", usecols=(1,2), delimiter=' ', unpack=True)
plt.figure(figsize=(8, 6))
plt.errorbar(x,f1,yerr=error1, label="Direct method")
plt.errorbar(x,f2,yerr=error2, label="Discretized method")
plt.xlabel('N blocks')
plt.ylabel('Put')
plt.axhline(y= 5.4595325819072364, linestyle="--", color="r", label="Theoretical value")
plt.title("Computation of the European put-option price")
plt.grid(True)
plt.legend()

plt.show()
```

### Commenti
Come si può osservare dai due grafici, entrambi i metodi di stima diretta e discretizzata convergono ai valori corrispondenti alla soluzione analitica di Black-Scholes con incertezze simili. Nel solo caso discreto della put-option il valore finale non rientra di $1\sigma$ nel valore analitico, ma lo fa entro $2\sigma$.

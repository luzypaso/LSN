```python
import matplotlib.pyplot as plt
import numpy as np
```

### TSP nel caso del cerchio

```python
dati = np.loadtxt("Coordinate_c.txt")

x = dati[:, 0]
y = dati[:, 1]
x = np.append(x, x[0])
y = np.append(y, y[0])

fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax = ax.flatten()

ax[0].plot(x, y, 'o-')
ax[0].set_xlabel("x")
ax[0].set_ylabel("y")
ax[0].set_title("Percorso migliore")
ax[0].grid(True)

gen, L_best, L_half = np.loadtxt("Costi_c.txt", unpack=True)

ax[1].plot(gen, L_best, label=r"$L^{(1)}$ miglior individuo")
ax[1].plot(gen, L_half, label=r"$\langle L^{(1)} \rangle$ metà migliore", alpha=0.7)
ax[1].set_xlabel("Generazioni")
ax[1].set_ylabel("Costo")
ax[1].set_title("Andamento funzione costo")
ax[1].grid(True)
ax[1].legend()

fig.suptitle("Città su una circonferenza", fontsize= 20)
plt.tight_layout()
plt.show()
```

### TSP nel caso del quadrato

```python
dati = np.loadtxt("Coordinate_q.txt")

x = dati[:, 0]
y = dati[:, 1]
x = np.append(x, x[0])
y = np.append(y, y[0])

fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax = ax.flatten()

ax[0].plot(x, y, 'o-')
ax[0].set_xlabel("x")
ax[0].set_ylabel("y")
ax[0].set_title("Percorso migliore")
ax[0].grid(True)

gen, L_best, L_half = np.loadtxt("Costi_q.txt", unpack=True)

ax[1].plot(gen, L_best, label=r"$L^{(1)}$ miglior individuo")
ax[1].plot(gen, L_half, label=r"$\langle L^{(1)} \rangle$ metà migliore", alpha=0.7)
ax[1].set_xlabel("Generazioni")
ax[1].set_ylabel("Costo")
ax[1].set_title("Andamento funzione costo")
ax[1].grid(True)
ax[1].legend()

fig.suptitle("Città dentro un quadrato", fontsize= 20)
plt.tight_layout()
plt.show()
print(f"Valore costo finale = {L_best[-1]}")
```

### Commenti
L'algoritmo si basa sulle classi Individual (ovvero un singolo percorso tra le città) e Population ($N_{pop} = 50$ individui). Ad ogni iterazione vengono estratti casualmente $N_{pop}$ individui e ognuno di essi ha una probabilità del 10% di mutare in ciascuno dei quattro modi possibili, così come ha una probabilità del 70% di realizzare un crossover con l'individuo successivo, in modo da dare due figli. Al termine di una generazione la popolazione dei figli diventa quella dei genitori e il ciclo si ripete.\
La selezione degli individui avviene estraendo casualmente un indice attraverso una distribuzione di probabilità che segue la legge di $x^3$ con $x \in [0,1)$.

Il primo grafico rappresenta il caso in cui tutte le _n città sono distribuite casualmente su una circonferenza. Con questa situazione banale il plot mostra la validità dell'algoritmo. Inoltre, come si può notare, la funzione costo si stabilizza dopo le prime $\sim 800$ generazioni.\
Il secondo grafico, invece, mostra che nel caso di città distribuite casualmente dentro un quadrato l'algoritmo è capace di trovare la soluzione migliore dopo le prime $\sim 1200$ generazioni.
